# Prior literature

**{cite:t}`Cagetti2005-tk`** — Journal of Political Economy
Volume 114, Number 5
October 2006 [DOI: 10.1086/508032](https://doi.org/10.1086/508032)


## Overview of the prior literature

This paper fills a gap in the literature by unifying incomplete-markets wealth models and entrepreneurship-with-frictions models in a quantitative general equilibrium framework. Earlier work either explained precautionary saving and moderate inequality (Aiyagari-type models) or showed that liquidity constraints affect entrepreneurial entry, but did not fully account for extreme top wealth concentration. This paper shows that when entrepreneurs face borrowing constraints yet have high-return, scalable technologies, wealth becomes a productive input, generating a self-reinforcing mechanism that endogenously produces large wealth inequality and links financial frictions to aggregate outcomes.

### Key Foundational Papers and Their Contributions

- **{cite:t}`Lucas1978-rd`** “On the Size Distribution of Business Firms”: Develops a general equilibrium model with heterogeneous managerial ability in which individuals choose between entrepreneurship and wage work. Firm size and the distribution of businesses arise endogenously from differences in talent and span-of-control technology.

- **{cite:t}`Evans1989-ws`** "An Estimated Model of Entrepreneurial Choice": Introduces liquidity constraints into a model of entrepreneurial entry and shows that personal wealth affects the probability of becoming an entrepreneur. Provides empirical and theoretical evidence linking wealth to business formation.

- **{cite:t}`Quadrini2000-mp`** "Entrepreneurship, Saving, and Social Mobility": Embeds entrepreneurs with borrowing constraints and idiosyncratic risk into an incomplete-markets macroeconomic model. Demonstrates that entrepreneurial risk and financial frictions generate high wealth concentration and strong precautionary savings.

- **{cite:t}`Kehoe1993-co`** "Debt-Constrained Asset Markets": Develops a general equilibrium model with limited contract enforcement that generates endogenous borrowing constraints. Shows how debt limits arise from incentive compatibility rather than exogenous restrictions.
