# Entrepreneurship, Frictions, and Wealth

## Summary

This paper constructs and calibrates a parsimonious model of occupational choice to study wealth inequality, entrepreneurial activity, and the role of financial frictions. Key findings include:

The main findings are listed below:

- **Motivation:** Wealth is much more concentrated than labor earnings in the data, and entrepreneurs are key to understanding top wealth holdings and capital accumulation. Earlier incomplete-markets models explained precautionary saving and moderate inequality but did not fully account for extreme top wealth concentration.
- **Unified framework:** The model allows for entrepreneurial entry, exit, and investment decisions in the presence of borrowing constraints (from limited contract enforcement). When entrepreneurs face binding borrowing limits but have high-return, scalable technologies, wealth becomes a productive input and a self-reinforcing mechanism generates large wealth inequality and links financial frictions to aggregate outcomes.
- **Fit and counterfactuals:** The model fits the observed wealth distribution for both entrepreneurs and workers very well. Counterfactuals show that tighter borrowing constraints reduce wealth concentration but also average firm size, aggregate capital, and the fraction of entrepreneurs; voluntary bequests allow high-ability workers to establish or enlarge firms and are important for the upper tail of the distribution.

## Non-Technical Methodical Overview

The paper unifies two strands of work in a quantitative general equilibrium framework:

1. **Incomplete-markets wealth models:** Agents face uninsurable risk and borrowing constraints, generating precautionary saving and moderate inequality (e.g. Aiyagari-type models), but these do not explain extreme top wealth concentration.

2. **Entrepreneurship-with-frictions models:** Liquidity and borrowing constraints affect who becomes an entrepreneur and how much they invest. Personal wealth affects the probability of entrepreneurial entry and firm size. The paper embeds this in a full equilibrium model where the distribution of wealth and the distribution of firm sizes are determined jointly.

3. **Wealth as a productive input:** When entrepreneurs have access to high-return, scalable technologies but cannot borrow freely, wealth becomes a productive input. Richer (or better-connected) households can run larger, more productive firms. This creates a self-reinforcing mechanism: successful entrepreneurs accumulate wealth, which relaxes their constraints and allows further investment, endogenously producing a thick right tail in the wealth distribution and linking financial frictions to aggregate capital, average firm size, and the fraction of entrepreneurs.

4. **Role of bequests:** The model includes altruism across generations (voluntary bequests). Voluntary bequests allow some high-ability workers to receive enough wealth to start or enlarge an entrepreneurial activity. With accidental bequests only, there would be fewer very large firms and less aggregate capital and wealth concentration.


## The Model

A parsimonious quantitative life-cycle model of occupational choice with heterogeneous agents, entrepreneurial entry and exit, and endogenous borrowing constraints.

### Features of the model

1. **Life cycle:** Agents live a finite number of periods. Each period they choose consumption and savings; they also make an occupational choice (worker vs. entrepreneur) and, if entrepreneurs, how much to invest in their firm.
2. **Borrowing constraints:** Debt repayment cannot be perfectly enforced, so borrowing limits arise endogenously from incentive compatibility (limited contract enforcement). This restricts how much entrepreneurs can leverage their wealth.
3. **Entrepreneurial technology:** Entrepreneurs have access to a productive technology whose return can depend on ability and scale; the technology is scalable but subject to the borrowing constraint, so wealth is a productive input.
4. **Heterogeneity:** Agents differ in productivity (e.g. entrepreneurial ability, labor productivity). These traits can be persistent or evolve over the life cycle and across generations.
5. **Bequests:** The model incorporates altruism across generations (voluntary bequests). At the end of life, agents leave bequests to their offspring, who start the next generation with that wealth. This links the distribution of wealth to the distribution of entrepreneurial opportunity.
6. **General equilibrium:** Wages and interest rates (and thus returns to saving and entrepreneurship) are determined in equilibrium. The distribution of wealth, the fraction of entrepreneurs, and the firm-size distribution are all endogenous.

### Preferences

Agents have standard time-separable preferences over consumption and, in some specifications, a warm-glow or altruistic motive for bequests (e.g. utility from leaving wealth to the next generation). Period utility is typically CRRA in consumption.

### Recursive formulation

The agent’s problem can be written recursively. The state includes current wealth (or assets), productivity, age, and possibly the current occupational status. The value function captures the maximum expected discounted utility from consumption and (if applicable) bequests. Choices each period are:

- Consumption and saving (subject to a no-borrowing or limited-borrowing constraint).
- Whether to be a worker (earn labor income, save at the risk-free rate or in a simple asset) or an entrepreneur (invest in the firm, earn entrepreneurial income subject to idiosyncratic risk and the borrowing constraint).

Entrepreneurial investment, firm size, and the decision to enter or exit entrepreneurship depend on current wealth and productivity. The solution yields policy functions for consumption, saving, and occupation; aggregation over the distribution of agents gives the equilibrium wealth distribution and firm-size distribution.

### Quantitative Analysis

The paper calibrates or estimates the model to match key empirical targets:

1. **External calibration:** Parameters such as the risk-free rate, labor income process, and the distribution of entrepreneurial productivity are set using external evidence or standard values.
2. **Moment matching:** Remaining parameters (e.g. borrowing limits, preference parameters, or details of the entrepreneurial technology) are chosen so that the model’s steady-state moments match the data—in particular the wealth distribution for workers and entrepreneurs, the fraction of entrepreneurs, and the distribution of firm sizes.
3. **Counterfactuals:** The calibrated model is used to study the effect of tighter or looser borrowing constraints, the removal of voluntary bequests (accidental bequests only), and related experiments on wealth concentration, aggregate capital, and entrepreneurial activity.




## Results

At the calibrated parameter values, the model fits the data very well along several dimensions:

- **Wealth distribution:** The model closely matches the observed wealth distribution for both entrepreneurs and workers, including the concentration of wealth at the top. The combination of borrowing constraints, scalable entrepreneurial technology, and voluntary bequests generates a right tail consistent with the data.
- **Entrepreneurial activity:** The equilibrium fraction of entrepreneurs and the distribution of firm sizes are in line with empirical observations, with wealth and ability determining who runs firms and at what scale.

### Counterfactuals

The paper uses the calibrated model to isolate the role of financial frictions and bequests:

1. **Tighter borrowing constraints:** More restrictive limits on borrowing reduce wealth concentration (less inequality) but also reduce average firm size, aggregate capital, and the fraction of entrepreneurs. This illustrates the trade-off between inequality and the level of entrepreneurial activity and capital accumulation.

2. **Accidental bequests only:** When voluntary bequests are shut down and only accidental bequests remain, there are fewer very large firms and less aggregate capital and wealth concentration. Voluntary bequests thus allow high-ability workers to establish or enlarge entrepreneurial activities and are important for matching the upper tail of the wealth distribution.

Together, these results show that the model successfully links financial frictions and bequest motives to both the cross-sectional distribution of wealth and to aggregate outcomes.



## Conclusion

Cagetti and De Nardi construct and calibrate a parsimonious model of occupational choice with entrepreneurial entry, exit, and investment under borrowing constraints. By unifying incomplete-markets wealth models and entrepreneurship-with-frictions models in a single general equilibrium framework, the paper shows that when entrepreneurs face binding borrowing limits but have access to high-return, scalable technologies, wealth becomes a productive input and a self-reinforcing mechanism generates large wealth inequality and links financial frictions to aggregate capital, firm size, and the fraction of entrepreneurs. The model fits the observed wealth distribution for entrepreneurs and workers very well. Counterfactuals demonstrate that tighter borrowing constraints reduce concentration but also reduce average firm size and aggregate capital, and that voluntary bequests are important for the upper tail. The framework has proven useful for subsequent work on wealth inequality, entrepreneurship, and policy.

## Limitations

The model is parsimonious by design; the following simplifications or extensions could matter for quantitative or policy work:

- **Taxation and transfers:** The paper does not incorporate progressive income or wealth taxation or detailed government transfer programs, which affect both occupational choice and the wealth distribution in the data.
- **Aggregate and cyclical risk:** The framework focuses on steady-state outcomes; aggregate shocks, business-cycle variation in credit, or time variation in the severity of financial frictions are not modeled.
- **Contract enforcement and collateral:** Borrowing limits are derived from limited contract enforcement in a reduced-form way; richer collateral or enforcement structures could alter the mapping from frictions to equilibrium.
- **Portfolio choice and asset heterogeneity:** Agents do not choose among multiple assets (e.g. housing, equity, private business) with different returns and liquidity; the single-asset structure keeps the model tractable but abstracts from portfolio composition and its role in wealth inequality.
- **Transition dynamics:** The main analysis is in steady state; transitional effects of policy changes or of changes in the borrowing regime are not studied.
